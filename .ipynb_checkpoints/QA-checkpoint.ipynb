```python
raw_datasets=load_dataset("squad",split="train[:10000]")
```

```python
train_size=int(0.8*len(raw_datasets))
val_size=len(raw_datasets)-train_size
```

```python
train_dataset_raw, val_dataset_raw = torch.utils.data.random_split(raw_datasets, [train_size, val_size])
```

```python
train_dataset_raw = Dataset.from_dict(train_dataset_raw[:])
val_dataset_raw = Dataset.from_dict(val_dataset_raw[:])
```

```python
print(f"Train dataset size: {len(train_dataset_raw)}")
print(f"Validation dataset size: {len(val_dataset_raw)}")
print("SQuAD dataset loaded.")
```

```python
print("\nLoading tokenizer...")
model_checkpoint = "bert-base-uncased" # A good starting point for QA
tokenizer = AutoTokenizer.from_pretrained(model_checkpoint)
print("Tokenizer loaded.")
```

```python
print("\nPreprocessing data for Question Answering...")

max_length = 384 # The maximum length of a feature (question and context)
doc_stride = 128 # The stride (overlap) between chunks of a long document

def preprocess_squad_examples(examples):
    # Tokenize questions and contexts
    # add_special_tokens=False to control special tokens manually for alignment
    # truncation="only_second" means only truncate the context if it's too long
    # return_offsets_mapping=True is crucial for mapping tokens back to original text
    # return_overflowing_tokens=True is needed to get overflow_to_sample_mapping when using stride
    questions = [q.strip() for q in examples["question"]]
    inputs = tokenizer(
        questions,
        examples["context"],
        max_length=max_length,
        truncation="only_second",
        stride=doc_stride,
        return_overflowing_tokens=True, # Ensure this is True for overflow_to_sample_mapping
        # Removed return_overflow_sample_ids=True as it caused TypeError
        return_offsets_mapping=True,
        padding="max_length",
    )

    # Map features to original examples
    # overflow_to_sample_mapping is returned when return_overflowing_tokens=True and stride is used
    sample_map = inputs.pop("overflow_to_sample_mapping")
    offset_mapping = inputs.pop("offset_mapping")

    # Find start and end positions of the answer in the tokenized input
    # Initialize lists for start and end token positions
    inputs["start_positions"] = []
    inputs["end_positions"] = []

    for i, offsets in enumerate(offset_mapping):
        sample_idx = sample_map[i]
        answer = examples["answers"][sample_idx]
        start_char = answer["answer_start"][0]
        end_char = start_char + len(answer["text"][0])

        # Sequence IDs (0 for question, 1 for context)
        sequence_ids = inputs.sequence_ids(i)

        # Find the start and end of the context in the tokenized sequence
        # The context starts after the [SEP] token following the question
        idx = 0
        while idx < len(sequence_ids) and sequence_ids[idx] != 1:
            idx += 1
        context_start_token = idx

        # The context ends before the [SEP] token at the end of the context
        idx = len(sequence_ids) - 1
        while idx >= 0 and sequence_ids[idx] != 1:
            idx -= 1
        context_end_token = idx

        # If the answer is not fully contained in this chunk, label as (0, 0)
        if not (offsets[context_start_token][0] <= start_char and offsets[context_end_token][1] >= end_char):
            inputs["start_positions"].append(0)
            inputs["end_positions"].append(0)
        else:
            # Otherwise, find the start and end token positions
            token_start_index = context_start_token
            # Corrected: Use token_start_index in offsets access
            while token_start_index < len(offsets) and offsets[token_start_index][0] <= start_char:
                token_start_index += 1
            inputs["start_positions"].append(token_start_index - 1)

            token_end_index = context_end_token
            # Corrected: Use token_end_index in offsets access
            while token_end_index >= 0 and offsets[token_end_index][1] >= end_char:
                token_end_index -= 1
            inputs["end_positions"].append(token_end_index + 1)

            # If the start position is greater than the end position, it's an invalid span.
            # This can happen if the answer text is slightly off or due to tokenization quirks.
            # In such cases, we set positions to 0 (representing the CLS token, often used for "no answer").
            if inputs["start_positions"][-1] > inputs["end_positions"][-1]:
                inputs["start_positions"][-1] = 0
                inputs["end_positions"][-1] = 0

    return inputs
```

```python
tokenized_train_dataset = train_dataset_raw.map(
    preprocess_squad_examples,
    batched=True,
    remove_columns=train_dataset_raw.column_names # Remove original columns
)
tokenized_val_dataset = val_dataset_raw.map(
    preprocess_squad_examples,
    batched=True,
    remove_columns=val_dataset_raw.column_names # Remove original columns
)
```

```python

print("Data preprocessing complete.")
print(f"Tokenized train dataset size: {len(tokenized_train_dataset)}")
print(f"Tokenized validation dataset size: {len(tokenized_val_dataset)}")
```

```python
print("\nLoading PyTorch model for Question Answering...")
# AutoModelForQuestionAnswering has two heads: one for start logits, one for end logits.
model = AutoModelForQuestionAnswering.from_pretrained(model_checkpoint)
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model.to(device)
print(f"PyTorch model loaded and moved to: {device}")

# --- 5. Define Training Arguments and Trainer ---
print("\nDefining TrainingArguments and Trainer...")
```

```python
def compute_metrics(eval_pred):
    # For Question Answering, the official evaluation script is complex.
    # We'll just return placeholder metrics here.
    # In a real scenario, you'd use the SQuAD evaluation script.
    return {"f1": 0.0, "exact_match": 0.0}
```

```python
training_args = TrainingArguments(
    output_dir='./qa_results',
    num_train_epochs=3, # Increased epochs for better learning
    per_device_train_batch_size=8, # A more reasonable batch size
    per_device_eval_batch_size=8,
    fp16=torch.cuda.is_available(), # Enable mixed precision if GPU is available
    warmup_steps=500,
    weight_decay=0.01,
    logging_dir='./qa_logs',
    logging_steps=50, # Log training loss every 100 steps
    eval_strategy="epoch", # Evaluate at the end of each epoch
    save_strategy="epoch", # Save checkpoint at the end of each epoch
    load_best_model_at_end=True, # Load the best model based on evaluation metric
    metric_for_best_model="eval_loss", # Use validation loss to determine best model
    greater_is_better=False, # Lower loss is better
    report_to="none", # Prevent external reporting
    # For a small dataset, you might not need a very aggressive learning rate
    learning_rate=2e-5,
    gradient_accumulation_steps=2
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_train_dataset,
    eval_dataset=tokenized_val_dataset,
    tokenizer=tokenizer,
    compute_metrics=compute_metrics, # This will be called during evaluation
)
print("Trainer initialized.")

# --- 6. Execute the training! ---
print("\nStarting fine-tuning of BERT for Question Answering...")
trainer.train()
print("Fine-tuning complete.")
```

```python
model_save_path = "./my_bert_qa_model"
tokenizer_save_path = "./my_bert_qa_model"
os.makedirs(model_save_path, exist_ok=True)
model.save_pretrained(model_save_path)
tokenizer.save_pretrained(tokenizer_save_path)
print(f"Fine-tuned model and tokenizer saved to {model_save_path}")

# --- 8. Perform Inference with the Fine-tuned Model using a Pipeline ---
print("\n--- Performing Inference with the Fine-tuned Model ---")

# Define pipeline_device here to ensure it's in scope for the pipeline creation
pipeline_device = 0 if torch.cuda.is_available() else -1 # Use GPU (device 0) if available, else CPU (-1)

# Load the fine-tuned model and tokenizer for the pipeline
qa_model = AutoModelForQuestionAnswering.from_pretrained(model_save_path)
qa_tokenizer = AutoTokenizer.from_pretrained(tokenizer_save_path)

# Create the QA pipeline
qa_pipeline = pipeline(
    "question-answering",
    model=qa_model,
    tokenizer=qa_tokenizer,
    device=pipeline_device # Use GPU if available, else CPU
)
```

```python
qa_examples = [
    {
        "question": "What is the capital of France?",
        "context": "France is a country located in Western Europe. Its capital and largest city is Paris. Paris is known for its art, fashion, gastronomy, and culture."
    },
    {
        "question": "When was the first iPhone released?",
        "context": "The first iPhone was unveiled by Steve Jobs on January 9, 2007, at the Macworld Conference & Expo in San Francisco. It went on sale in the United States on June 29, 2007."
    },
    {
        "question": "What is the main purpose of the human heart?",
        "context": "The human heart is a muscular organ that pumps blood throughout the body. It is the primary organ of the circulatory system."
    },
    {
        "question": "Who developed the Python programming language?",
        "context": "Python is a high-level, general-purpose programming language. Its design philosophy emphasizes code readability with the use of significant indentation. Python was conceived in the late 1980s by Guido van Rossum at Centrum Wiskunde & Informatica (CWI) in the Netherlands as a successor to the ABC language."
    }
]

print("\nRunning QA pipeline on sample examples:")
for example in qa_examples:
    result = qa_pipeline(question=example["question"], context=example["context"])
    print(f"Context: {example['context']}")
    print(f"Question: {example['question']}")
    print(f"Answer: {result['answer']} (Score: {result['score']:.4f})\n")

# --- 9. Plotting Training Loss (from Trainer logs) ---
print("\nPlotting training and validation loss...")
log_history = trainer.state.log_history
train_losses = []
eval_losses = []
steps = []

for log in log_history:
    if 'loss' in log:
        train_losses.append(log['loss'])
        steps.append(log['step'])
    if 'eval_loss' in log:
        eval_losses.append(log['eval_loss'])

# Ensure eval_losses match steps (evals happen at epoch end)
eval_steps = [log['step'] for log in log_history if 'eval_loss' in log]

if train_losses:
    plt.figure(figsize=(10, 6))
    plt.plot(steps, train_losses, label='Training Loss', marker='o', linestyle='-', color='skyblue', markersize=4)
    if eval_losses:
        plt.plot(eval_steps, eval_losses, label='Validation Loss', marker='x', linestyle='--', color='salmon', markersize=6)
    plt.title('Training and Validation Loss over Steps')
    plt.xlabel('Training Steps')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    plt.tight_layout()
    plt.show()
else:
    print("No training loss data found in log history to plot.")

print("\nExtractive QA project guide complete!")
```
